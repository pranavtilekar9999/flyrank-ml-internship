## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
## 1) My lane as an ML task

# My selected lane is Lane 2: Refresh / Content Opportunity Scoring.

# The ML task type is **scoring**.

# The goal is to assign each page a score representing its priority for content review. Pages can then be ranked and mapped to actions such as refresh, expansion, protection, pruning, or monitoring.

# The unit being scored is an individual content page.
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
## 2) Target or proxy

# For the starter dataset, I will use `is_declining_label` as a proxy target.

# The label indicates whether a page is currently classified as declining based on its observed trend direction.

# This is useful for the starter exercise because it gives a concrete signal that can be modeled. However, it is not an ideal future-looking target for content refresh decisions because it describes an observed condition rather than directly measuring whether a future refresh would improve the page.

# For a future version of the project, I would want a target that better represents the outcome of interest, such as whether a page becomes a successful refresh candidate based on subsequent performance.
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
## 3) Success metric

# Because the output is a ranked review queue, I would evaluate the quality of the highest-ranked pages rather than only looking at overall accuracy.

# A useful metric is **Precision@K**, for example Precision@100.

# This measures how many of the top 100 pages selected by the scoring system actually match the proxy target.

# This matters because the practical goal is to help a content team decide which pages deserve attention first. A useful scoring system should place a high proportion of relevant pages near the top of the queue.
```
